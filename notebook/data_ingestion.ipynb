```python
# ================================
# Configuration
# ================================

from pathlib import Path

# Project Information
PROJECT_NAME = "Crypto Volatility Prediction"
PROJECT_VERSION = "1.0"

# Project Directories
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"
RAW_DATA_DIR = DATA_DIR / "raw"
PROCESSED_DATA_DIR = DATA_DIR / "processed"

# Dataset
DATASET_NAME = "data.csv"
DATASET_PATH = RAW_DATA_DIR / DATASET_NAME

# Dataset Metadata
TARGET_COLUMN = "volatility"
DATE_COLUMN = "date"
ID_COLUMN = None          # Example: "customer_id" or "transaction_id"

# Reproducibility
RANDOM_STATE = 42

print("Configuration Loaded Successfully!")
```

```python
# ================================
# Dataset Loading
# ================================

try:
    df = pd.read_csv(DATASET_PATH)
    print("Dataset loaded successfully.")
    if not df.empty:
     print(f"Dataset Shape: {df.shape}")
    else:
     print("Dataset loaded, but it contains no rows.")

except FileNotFoundError:
    print(f"Error: File not found at {DATASET_PATH}")

except pd.errors.EmptyDataError:
    print("Error: The dataset is empty.")

except pd.errors.ParserError:
    print("Error: Unable to parse the dataset. Please check the file format.")

except Exception as e:
    print(f"An unexpected error occurred: {e}")





```

```python
EXPECTED_COLUMNS = [
    "date",
    "crypto_name",
    "open",
    "high",
    "low",
    "close",
    "volume",
    "marketCap"
]
```

```python
print(df.info())
COLUMNS =[col for col in df.columns]
print(COLUMNS)
# missing columns 
missing_columns = set(EXPECTED_COLUMNS) - set(COLUMNS) 
print(f"missing_columns:{missing_columns}")
#extra columns
extra_columns = set(COLUMNS) - set(EXPECTED_COLUMNS)
print(f"Extra Columns: {extra_columns}")
#schema check
if set(EXPECTED_COLUMNS).issubset(set(COLUMNS)):
    print("Schema Validation Passed")
else:
    print("Schema Validation Failed") 
#duplicates
if df.columns.duplicated().any():
    print("Duplicate column names found.")
else:
    print("No duplicate column names.")
        
```

```python
#data type validation 
EXPECTED_DTYPES = {
    "open": "float64",
    "high": "float64",
    "low": "float64",
    "close": "float64",
    "volume": "float64",
    "marketCap": "float64",
    "timestamp": "object",
    "crypto_name": "object",
    "date": "object"
}
DTYPES = df.dtypes.to_dict()
mismatch = []
for col, expected_dtype in EXPECTED_DTYPES.items():
    if col in DTYPES:
     actual_dtype = str(DTYPES[col])
    else:
     print(f"{col} is missing.")
    if expected_dtype == actual_dtype:
      print(f"{col} validated")
    else:
      print(f"{col} not validated")
      mismatch.append({
          "column" :col,
          "expected": expected_dtype,
          "actual":actual_dtype 
      })

if not mismatch:
    print("\n All data types are valid.")
else:
    print("\n Data type mismatches found:")
    for item in mismatch:
        print(item)
     

```

```python
#data integrity validation 
#missing values 
missing_values=df.isnull().sum()
if missing_values.sum() == 0:
     print("No missing values found.")
else:
     print("Missing values:")
     print(missing_values[missing_values > 0])

#duplicates rows 
duplicate_rows=df.duplicated().sum()
if duplicate_rows == 0:
    print(" No duplicate rows found.")
else:
    print(" duplicate rows found.")

#constant column >> columns with only one values
df.nunique()
constant_columns = []

for col in df.columns:
    if df[col].nunique() == 1:
        constant_columns.append(col)
if not constant_columns:
    print(" No constant columns.")
else:
    print(constant_columns)



```
